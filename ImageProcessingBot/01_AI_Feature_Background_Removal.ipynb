## 1. Setup and Imports

```python
# Cell 1 - Installing Libraries and Importing Packages
%pip install rembg[gpu] gradio --quiet

import gradio as gr
from PIL import Image
from rembg import remove, new_session
import os

print("Libraries for Background Removal are ready.")
```

## 2. AI Model Loading

```python
# Cell 2 - AI Model Setup
# This points the U2NET_HOME environment variable to our local models folder.
# 'rembg' will now look for and load models from here.
os.environ["U2NET_HOME"] = "./saved-models"

print("Preparing model sessions from local './saved-models' directory...")
# Pre-load the session for the anime model for faster performance.
anime_session = new_session("isnet-anime")
print("Sessions are ready.")
```

## 3. AI Feature Definition

```python
# Cell 3 - AI Feature Definition

def remove_background_feature(input_image: Image.Image, model_choice: str) -> Image.Image:
    """
    Takes a PIL Image and a model choice, removes the background, 
    and returns the processed image.
    """
    if input_image is None:
        return None
    
    # Use the appropriate session based on the user's choice from the dropdown
    if model_choice == "Anime (isnet-anime)":
        return remove(input_image, session=anime_session)
    else: # Default to the general-purpose model
        return remove(input_image)
```

## 4. Gradio User Interface

```python
# Cell 4 - Gradio User Interface for Background Removal

with gr.Blocks(theme=gr.themes.Soft()) as iface:
    gr.Markdown("# 🤖 AI Feature: Background Removal")
    gr.Markdown("Upload an image to remove its background. Choose the 'Anime' model for illustrations and cartoons for best results.")
    
    with gr.Row():
        with gr.Column(scale=2):
            input_img = gr.Image(type="pil", label="Upload Your Image")
            model_choice_dd = gr.Dropdown(
                ["General (u2net)", "Anime (isnet-anime)"], 
                label="Select Model", 
                value="General (u2net)",
                info="Choose 'Anime' for illustrations for best results."
            )
            submit_btn = gr.Button("Remove Background", variant="primary")
            
        with gr.Column(scale=3):
            output_img = gr.Image(type="pil", label="Processed Image")

    submit_btn.click(
        fn=remove_background_feature,
        inputs=[input_img, model_choice_dd],
        outputs=output_img
    )

# Launch the interface
iface.launch(share=True)
```
